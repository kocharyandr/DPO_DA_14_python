#### Всё вместе:

```python
import requests
from bs4 import BeautifulSoup

url_cern = 'http://info.cern.ch/hypertext/WWW/TheProject.html' # первый веб-сайт

page = requests.get(url_cern)
#print("Код состояния: ", page.status_code)

page.encoding = 'utf-8'

soup = BeautifulSoup(page.text)

print(soup.prettify())
```

#### Теперь можно выбирать элементы по характеристикам: тегу, классу...

```python
links = soup.find_all('a') # например, можно найти все ссылки (тег <a>)
links # это список, по нему можно пройти циклом!
```

```python
links[0] # самая первая ссылка
```

```python
links[0].text # текст ссылки
```

```python
links[0].get('href') # куда ведёт ссылка
```
