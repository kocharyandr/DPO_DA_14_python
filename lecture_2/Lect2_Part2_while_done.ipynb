## Цикл `while`

**While** - один из самых универсальных циклов в Python, поэтому достаточно медленный.

Цикл **while** (в переводе - **пока**) позволяет выполнить одну и ту же последовательность действий, **пока** проверяемое условие истинно

<img src="https://i.ibb.co/w0dP6n0/Untitled-Frame-1-15.jpg">

**Зачем**:

Как правило, цикл while используется, когда невозможно определить точное значение количества проходов исполнения цикла.

**Условие**:

Условие записывается до тела цикла и проверяется до выполнения тела цикла.

**Еще раз подробно**:

При выполнении цикла **while** сначала проверяется условие. Если оно ложно, то выполнение цикла прекращается и управление передается на следующую инструкцию после тела цикла **while**. Если условие истинно, то выполняется инструкция, после чего условие проверяется снова и снова выполняется инструкция. Так продолжается до тех пор, пока условие будет истинно. Как только условие станет ложно, работа цикла завершится и управление передастся следующей инструкции после цикла.

Синтаксис цикла **while** в простейшем случае выглядит так:

````python
while <<условие>> :
    <<блок инструкций>>
````

### Задача № 1

Попробуем вывести все числа от 1 до заданного пользователем целого числа N с помощью цикла while

````python
N = int(input()) # пользователь вводит число
print('')


i = 1            # вводим вспомогательный счетчик

while i < N:    # проверяемое условие
    print(i)     # вывод числа на экран в теле цикла
    i = i + 1    # переходим к следующему числу, увеличиваем счетчик на 1 (Вариант 2: i+=1)


print('')
# асмоистязание программиста
# N = 5
i = 1
if i < N:
  print(i)
  i += 1 # i = i + 1
if i < N:
  print(i) # 2
  i += 1 # 3

if i < N:
  print(i) # 3
  i += 1 # 4

if i < N:
  print(i) # 4
  i += 1 # 5
if i < N: # 5 < 5 = False
  print(i) # 2
  i += 1 # 3
````

В этом примере переменная i внутри цикла изменяется от 1 до N. Такая переменная, значение которой меняется с каждым новым проходом цикла, называется счетчиком

### Задача № 2
Теперь попробуем посчитать из скольки цифр состоит натуральное число, введенное пользователем с клавиатуры с помощью цикла while

````python
N = int(input())    # пользователь вводит число

length = 0          # вводим вспомогательный счетчик

while N > 0:        # проверяемое условие
    N //= 10        # целочисленное деление на 10
    length += 1     # переходим к следующему числу, увеличиваем счетчик на 1

print(length)       # выводим результат
````

### Задача № 4
Программа по вычислению факториала числа, заданного пользователем

!n = n * (n - 1) * (n - 2) * (n -3) ... 3 * 2 * 1

!4 = 4 * 3 * 2 * 1

````python
n = int(input())

factorial = 1

history = ''
while n > 0:
  factorial *= n # factorial = factorial * n
  history += ' * ' + str(n)
  n -= 1 # n = n - 1
  print(f'выражение: {history[3:]}, n = {n}')

print(factorial)
````

Осталось написать цикл. До тех пор, пока пользователь не угадает число, программа не будет останавливаться, но зато она будет давать подсказки: если введенное пользователем число больше загаданного, то будет выводиться сообщение "Вы ввели слишком большое число.", если меньше – "Вы ввели слишком маленькое число."

````python
while True:
    guess = int(input("Ваша попытка: "))
    if guess == n:
        print("Вы выиграли!")
        break
    elif guess > n:
        print("Вы ввели слишком большое число.")
    else:
        print("Вы ввели слишком маленькое число.")
````

Казалось бы, зачем нам нужно еще и `else`, если действие можно так же прописать и после выхода из цикла, а не на выходе? Это имеет смысл в комбинации с инструкцией `break`. После ее исполнения цикл прерывается полностью, причем часть с `else` не выполняется. Вспомним нашу задачку с поиском Николая. Может быть и так, что вы неожиданно встретите свою подругу, Маргариту. При этом вы так обрадуетесь, что даже забудете на время про поиск Николая и погрузитесь в беседу с ней. Давайте покажем это в коде:



```

while <<условие>> :
    <<блок инструкций>>
else:
   <<блок инструкций>>
```



````python
index = 0 # начинаете с первого посетителя в списке

while visitors[index] != 'Николай': # продолжать, пока не увидите Николая
    if visitors[index] == 'Маргарита': # прописываем условие на случай встречи Маргариты
        print('Ух ты, это же Маргарита! Как давно я ее не видел... Надо подойти поболтать.')
        break # отвлеклись на Маргариту и ВЫХОДИМ ИЗ ЦИКЛА (если встретили)
    else: # не встретили ни Колю, ни Маргариту; ищем дальше
        print(f'Беда, это не Николай. Этот {index + 1}-й по счету человек - {visitors[index]}.') # смотрим на порядок и имя
        # человека
        index += 1 # index += 1 это то же самое, что index = index + 1 (увеличиваем наш индекс на единицу, чтобы посмотреть на
                    # следующего в очереди человека)
else: # сюда не доходим никогда, потому что прежде Николая нам встретилась коварная Маргарита
    print()
    print('Ура, а вот и он! Наконец-то! Теперь можно и в бар.')
````

Стоит упомянуть еще и об инструкции `continue`. После ее исполнения начинается следующая итерация цикла, минуя все действия, которые остались в текущей. Немного изменим нашу ситуацию. Представим, что некоторое время назад вы поругались с Маргаритой, и теперь стараетесь вообще о ней не думать. Тогда, если ваш взгляд натолкнется на нее, вы даже не подумаете "Беда, это не Николай...", а просто переведете глаза на следующего человека. Тут-то нам и поможет инструкция `continue`:

````python
index = 0 # начинаете с первого посетителя в списке

while visitors[index] != 'Николай': # продолжать, пока не увидите Николая
    if visitors[index] == 'Маргарита': # прописываем условие на случай встречи Маргариты
        index += 1 # сразу идем дальше
        print('*это Маргарита, но мы обиделись, мы не будем о ней думать*')
        continue # отвлеклись на Маргариту, но НЕ ДУМАЕМ о ней (поссорились)
    print(f'Беда, это не Николай. Этот {index + 1}-й по счету человек - {visitors[index]}.') # смотрим на порядок и имя человека
    index += 1 # index += 1 это то же самое, что index = index + 1 (увеличиваем наш индекс на единицу, чтобы посмотреть на
                    # следующего в очереди человека)
else:
    print()
    print('Ура, а вот и он! Наконец-то! Теперь можно и в бар.')
````

## Вложенные циклы


Вложенные циклы `while` содержат два основных компонента:

1. Внешний цикл `while`
2. Внутренний цикл `while`


Внешний цикл `while` может содержать несколько внутренних циклов `while`. Условие приводит к булевому значению, и если оно истинно, только тогда цикл выполняется. На каждой итерации внешнего цикла `while` внутренний цикл выполняется с самого начала, и этот процесс продолжается до тех пор, пока условие внешнего цикла не станет истинным. Аналогично, внутренний цикл `while` выполняется только в том случае, если его условие истинно, после чего выполняется блок кода.

### Задача № 2

Выведите таблицу умножения от 1 до 5.

````python
i = 1
while i<=5:
  # i*j
  j = 1
  while j <= 5:
    print(f'{i} * {j} = {i * j}', end = ' | ')
    j += 1
  i+=1
  print()
````
