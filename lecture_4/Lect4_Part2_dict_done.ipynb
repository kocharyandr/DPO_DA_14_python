✍️ _**Словарь**_ — это изменяемый и неупорядоченный тип данных, где каждому уникальному ключу сопоставлено некоторое значение.

🔑 Ключи и значения
* **Ключи** — неизменяемые типы (строки, числа, кортежи без изменяемых объектов).
* **Значения** — любые объекты.

🚨 Важные особенности
* Порядок с Python 3.7 сохраняется (в порядке добавления элементов).
* Изменяемость: можно добавлять и удалять элементы.
* Ключи уникальны — при повторе старое значение перезаписывается.

### Добавить новое значение в словарь

<div style="background-color: rgba(244, 194, 194, 0.31); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(244, 194, 194, 0.57);">

🧩 Добавление нового элемента в словарь

Чтобы добавить новую пару ключ–значение в словарь, используется оператор присваивания `[]`.
Если ключа ещё нет — он будет добавлен.

📌 Синтаксис
```
dictionary[new_key] = new_value
```
* `new_key` — ключ, по которому будет доступно значение.
* `new_value` — само значение, связанное с этим ключом.

🔍 Важные особенности
* Если ключ уже есть — значение перезапишется.
* Ключом может быть любой хешируемый объект. Чаще всего — строки, числа или кортежи.
* Можно добавлять данные и в пустой словарь.

````python
slovar = {}
slovar ['asda']= ['asdas']
````

### Удалить элемент

<div style="background-color: rgba(244, 194, 194, 0.31); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(244, 194, 194, 0.57);">

🗑 **Удаление элемента из словаря**

В Python существует несколько способов удалить элемент из словаря по ключу.
Это методы `.pop()` и оператор `del`.

*🔹 Способ 1 — Метод `.pop()`
Метод `.pop()` удаляет элемент по ключу и возвращает его значение.

📌 Синтаксис
```
dictionary.pop(key[, default])
```
* `key` — ключ, который нужно удалить.
* `default` (необязательный) — значение, возвращаемое, если ключ отсутствует. Если не указан — будет KeyError.

*🔹 Способ 2 — Оператор `del`
Удаляет пару ключ–значение без возврата значения.

📌 Синтаксис
```
del dictionary[key]
```

````python
dct = dict(cat='кошка', dog="собака")
dct
````

````python
print(dct.pop('cat', 'нет в словаре'))
print(dct)
````
