# Задача 1: ось симметрии

Даны n точек на плоскости. Определите, существует ли вертикальная прямая (параллельная оси Y), относительно которой все точки симметричны. Иначе говоря, если отразить все точки относительно этой прямой, множество точек не изменится. Точки могут повторяться.


points = [[1, 1], [-1, 1]]
→
True

points = [[1, 1], [-1, -1]]
→
False

```python
# .     |     .
#    .  |  .
#       |
#       |

# a.    s.    b
# -5    0     5

# -5 - 0 = -5
#  0 - (-5) = 5
# d = dist( x_s, x_a)
# x_b = x_s ± d


def search_s(points):
  # points = [[x_0,y_0],[x_1,y_1], ..., [x_n,y_n]]

  min_x, max_x = float("-inf"), float('inf')
  points_set = set()
  x_s = None
  for x, y in points:


    min_x = min(x, min_x)
    max_x = max(x, max_x)
    points_set.add((x, y))

  x_s = (min_x + max_x)/2 # Знаю x_0 как найти x_1? Ответ: x_1 = 2 * x_s - x_0
# просто выразить из уравнения выше
# 2 * x_s = x + x_sim -> x_sim = 2 * x_s - x
  for x, y in points:
    x_sim = x_s * 2 - x
    if (x_sim, y) not in points_set:
        return False

  return True


```

# Задача № 3: сумма 2 чисел

Дан массив целых чисел nums и число target. Верните индексы двух чисел, сумма которых равна target. Гарантируется, что решение ровно одно. Один и тот же элемент нельзя использовать дважды.




nums = [2, 7, 11, 15], target = 9
→[0, 1]


nums = [3, 2, 4], target = 6
→[1, 2]


nums = [3, 3], target = 6
→[0, 1]

```python
def find_targer(nums, target):

  for i in range(len(nums)):
    x_0 = nums[i]
    for j in range(i + 1, len(nums)):
      if x_0 + nums[j] == target:
        return i,j
  return -1

def optimized(nums, target):
  # Пусть target = x_0 + x_1
  # x_0 есть, а вот есть ли его пара x_1?
  # x_1 есть массив или нет

  data = {} # в ключе храним само значения, в values индекс

  # for i in range(len(nums)):
  #   x_0 = nums[i]
  for i, x in enumerate(nums):
    if target - x in data:
      return i, data[target - x]
    data[x] = i

  return -1
```
