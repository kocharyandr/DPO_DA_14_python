Что такое **«экранирование»**?

Обратный слеш `\` в Python — это как волшебный «ключ», который говорит:

«Внимание! Следующий символ нужно обрабатывать не как обычный текст, а как специальный».

Такой «спец-символ» называется `экранированным` (escaped), потому что он «вырывается» из обычного потока строк и приобретает особый смысл.

**Пример:**

В Python вместо `«↵»` пишем `\n`.

`\` — это знак, который предупреждает: «готовься к спец-символу».

`n` — это буква, обозначающая именно new line (новая строка).

**Другие популярные экранированные символы:**

* `\t` — табуляция (отступ)

* `\'` — одинарная кавычка внутри одинарной строки

* `\"` — двойная кавычка внутри двойной строки

* `\\` — сам обратный слеш

Пример с экранированием и переносом строки:

````python
print('\tПривет!\n\tРад слышать, что у тебя все хорошо,\nдруг по переписке.\nРад слышать, что у тебя все хорошо, \nРад слышать, что у тебя все хорошо, \n\tP.S. У меня тоже все хорошо!')
````

Также можно извлекать несколько символов с помощью срезов.

**Срез** позволяет извлекать фрагменты строки, задавая диапазон индексов.

_Синтаксис:_

```
substring = s[start:stop]
```
* `start` — индекс первого символа во фрагменте (включительно). Если опущен, по умолчанию 0.

* `stop` — индекс символа после конца фрагмента (не включительно). Если опущен, по умолчанию len(s).

Также может быть срез с шагом.

_Синтаксис:_

```
substring = s[start:stop:step]
```

* `step` — шаг (через сколько символов “перескакивать”). Если опущен, по умолчанию 1.

````python
s = "Programming"
     012345678910
s[0:3]
````

````python
s[::-1]
````

````python
s = "Programming"
print(s[3:8])   # 'gramm' — символы с индексами 3,4,5,6,7
````

````python
s = "Сегодня прекрасный день"
print(s[1:9:2])  # 'bdfh'  (индексы 1,3,5,7)
print(s[::3])    # 'adgj'  (от начала до конца, каждый 3-й)
print(s[::-1])   # 'jihgfedcba' — разворот строки
````
