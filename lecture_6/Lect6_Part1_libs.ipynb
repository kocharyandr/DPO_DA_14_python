# Библиотеки и модули

Мы уже умеем пользоваться встроенными функциями Python и создавать собственные функции, но возможности языка на этом не заканчиваются. Теперь разберёмся, что такое библиотеки и модули и как их использовать.

Если несколько функций решают похожие задачи, их объединяют в модули. Например, модуль <code style="color:red;background:none">math</code> содержит математические функции, модуль <code style="color:red;background:none">string</code> предназначен для работы со строками, а модуль <code style="color:red;background:none">random</code> предоставляет функции для генерации псевдослучайных чисел. В модуле могут храниться не только функции, но и константы и другие данные.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Модуль</b> — это файл с кодом на Python (обычно с расширением <code style="color:red;background:none">.py</code>), внутри которого находятся функции, переменные, классы и другие объекты, объединённые одной задачей или тематикой. </div>

## Пример использования модуля стандартной библиотеки

Представим, что мы хотим сделать квадратную лужайку и засадить её искусственным газоном. Газон продаётся квадратами размером 1 метр на 1 метр; мы купили 16 таких квадратов. Чему будет равна длина стороны лужайки?
​

Из школьной математики: длина стороны квадрата равна квадратному корню из его площади. В нашем случае площадь равна 16, значит, нужно вычислить корень из 16. Корень из 16 равен 4, потому что <code style="color:red;background:none">4 * 4 = 16</code>.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Идея:</b> вместо того чтобы считать корень вручную, мы напишем программу на Python, которая сможет делать это автоматически, даже для больших чисел. </div>
Для вычисления квадратного корня воспользуемся функцией <code style="color:red;background:none">sqrt()</code> из модуля <code style="color:red;background:none">math</code>. Эта функция как раз вычисляет квадратный корень числа.
​

```python
sqrt(16) # Пытаемся вызвать функцию sqrt() из модуля math
```

Не получилось. Потому что в отличие от тех частых, нужных, популярных функций типа `print()`, которые мы уже использовали, функция `sqrt()` "убрана" в отдельный модуль — и этот модуль надо **импортировать**. 

```python
import math # Импортируем модуль math
```

```python
sqrt(16) # Снова пытаемся вызвать функцию sqrt() из модуля math, предварительно его импортировав
```

Снова не получилось. Дело в том, что мы никак не пояснили, что это _именно та_ функция `sqrt()`, которая лежит в модуле `math`, а не какая-то другая. Чтобы это указать, нужно написать через точку: `<модуль>.<функция()>`. Попробуем.

```python
math.sqrt(16) # снова пытаемся вызвать функцию sqrt(), указав, что это функция из модуля math
```

Получилось!

Теперь посчитаем сторону квадратной лужайки для случая с 4489 квадратиками.

```python
x = math.sqrt(4489)
```

```python
print(x)
```

Длина стороны равна 67 метрам — большая будет лужайка!

## Импорт отдельных компонентов модуля

Иногда неудобно каждый раз писать название модуля, точку и название функции, например <code style="color:red;background:none">random.randint()</code> или <code style="color:red;background:none">math.sqrt()</code>.


Чтобы сократить запись, мы можем явно указать, какие именно объекты хотим импортировать из модуля, и обращаться к ним напрямую, без префикса с именем модуля.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Синтаксис:</b><br> <code style="color:red;background:none">from random import randint, choice</code> — после такого импорта можно вызывать <code style="color:red;background:none">randint()</code> и <code style="color:red;background:none">choice()</code> без <code style="color:red;background:none">random.</code>. </div>
Рассмотрим это на примере двух функций из модуля <code style="color:red;background:none">random</code>, которые позволяют генерировать псевдослучайные числа.

```python
from random import randint, uniform # Импортируем две функции из модуля random, здесь скобки не нужны
```

Чтобы узнать, что делают эти функции и какие аргументы им нужно подавать, можно посмотреть документацию или воспользоваться функцией `help()`:

```python
help(randint)
```

```python
help(uniform)
```

Мы видим всю информацию о функции `randint()`: она содержится в модуле `random`, возвращает целое псевдослучайное число в заданном диапазоне, принимая на вход границы диапазона.

Аналогичным образом можно узнать, что функция `uniform()` принимает границы диапазона и возвращает дробное псевдослучайное число.

Посмотрим, как сработают эти функции в диапазоне от 100 до 200:

```python
# Вызываем функции по 5 раз
print([randint(100, 200) for _ in range(5)])
print([uniform(100, 200) for _ in range(5)])
```

Посмотреть на документацию можно не только у функций, но и у методов:

```python
help(str.upper)
```

# Импорт под псевдонимом
Иногда имя модуля оказывается длинным или часто повторяется в коде, и его неудобно всё время писать целиком. В этом случае можно дать модулю более короткое имя — псевдоним.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Синтаксис:</b><br> <code style="color:red;background:none">import string as s</code> — теперь вместо <code style="color:red;background:none">string</code> в коде можно писать <code style="color:red;background:none">s</code>. </div>
Рассмотрим пример с модулем <code style="color:red;background:none">string</code>, предназначенным для работы со строками. Допустим, мы решили, что слово <code style="color:red;background:none">string</code> писать слишком долго и хотим использовать короткое имя <code style="color:red;background:none">s</code>.

```python
import string as s    # Даем псевдоним модулю string
print(s.punctuation)  # Печатаем строку со знаками препинания, хранящуюся в модуле 
```

```python
s.ascii_uppercase
```

Теперь мы вместо `string.punctuation` будем писать `s.punctuation`, а `string.punctuation` не сработает.

```python
print(string.punctuation)
```

## Пример модуля: pprint
Модуль <code style="color:red;background:none">pprint</code> позволяет «красиво» и наглядно выводить сложные объекты Python: словари, списки списков, вложенные структуры и т.п. Он автоматически форматирует вывод, добавляет переносы строк и отступы, чтобы структуру данных было легче читать.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Документация:</b><br> Официальное описание и примеры использования: <a href="https://docs.python.org/3/library/pprint.html ">https://docs.python.org/3/library/pprint.html</a> </div>

```python
from pprint import pprint

s = [['Греция', 'Афины', 131_940, 10_793_526],['Грузия', 'Тбилиси', 69_700, 3_720_400],['Дания', 'Копенгаген', 43_094, 5_707_251]]

print(s)
print("\n-----------------------------\n")
pprint(s)
```

```python
m = {'Европа': [['Греция', 'Афины', 131_940, 10_793_526], ['Дания', 'Копенгаген', 43_094, 5_707_251]], 'Азия': [['Грузия', 'Тбилиси', 69_700, 3_720_400]]}

pprint(m, depth=1) # показать 1 уровень вложенности
print("\n-----------------------------\n")
pprint(m, depth=2) # показать 2 уровня вложенности
print("\n-----------------------------\n")
pprint(m, depth=3) # показать 3 уровня вложенности
print("\n-----------------------------\n")

# pprint сортирует словари при выводе, что не всегда удобно. Но сортировку можно отключить.
pprint(m, depth=2, sort_dicts=False)
```
