```python
def get_page(p):
    url = f'http://books.toscrape.com/catalogue/page-{p}.html'
    response = requests.get(url)
    tree = BeautifulSoup(response.content, 'html.parser')
    books = tree.find_all('article', {'class' : 'product_pod'})
    
    info = []
    
    for book in books:
        info.append({'price': book.find('p', {'class': 'price_color'}).text,
                     'href': book.h3.a.get('href'),
                     'title': book.h3.a.get('title'),
                    'rating': book.p.get('class')[1]})
                     
    return info
```

Напишите программу, где пользователь вводит тег, а ему выводятся все цитаты, у которых есть такой тег (например, 'truth').

### Решение

Посмотрим на одну цитату: каждая лежит в `<div class="quote">`, текст — в `<span class="text">`, автор — в `<small class="author">`, теги — в ссылках `<a class="tag">`.

```python
import requests
import time
import pandas as pd
from bs4 import BeautifulSoup

url = 'https://quotes.toscrape.com/page/1/'
response = requests.get(url,timeout=5)
tree = BeautifulSoup(response.content, 'html.parser')

quotes = tree.find_all('div', {'class': 'quote'})
quotes[0]
```

Соберём функцию для одной страницы по аналогии с `get_page` для книг. Сколько всего страниц, заранее не знаем, поэтому функция заодно проверяет, есть ли кнопка *Next* (`<li class="next">`).

```python
def get_quotes_page(p):
    url = f'https://quotes.toscrape.com/page/{p}/'
    response = requests.get(url)
    tree = BeautifulSoup(response.content, 'html.parser')
    quotes = tree.find_all('div', {'class': 'quote'})

    info = []

    for quote in quotes:
        info.append({'text': quote.find('span', {'class': 'text'}).text,
                     'author': quote.find('small', {'class': 'author'}).text,
                     'tags': [tag.text for tag in quote.find_all('a', {'class': 'tag'})]})

    # есть ли ссылка на следующую страницу
    has_next = tree.find('li', {'class': 'next'}) is not None
    return info, has_next
```

```python
all_quotes = []
p = 1

while True:
    info, has_next = get_quotes_page(p)
    all_quotes.extend(info)
    if not has_next:
        break
    p += 1
    time.sleep(1)

print(len(all_quotes), 'цитат с', p, 'страниц')
```

Теперь сама программа: пользователь вводит тег, а мы выводим все цитаты, у которых есть такой тег.

```python
def find_by_tag(quotes, tag):
    return [q for q in quotes if tag in q['tags']]

tag = input('Введите тег: ').strip().lower()
result = find_by_tag(all_quotes, tag)

if result:
    print(f'Цитат с тегом "{tag}": {len(result)}\n')
    for q in result:
        print(q['text'])
        print('—', q['author'])
        print()
else:
    print(f'Цитат с тегом "{tag}" не найдено')
```
