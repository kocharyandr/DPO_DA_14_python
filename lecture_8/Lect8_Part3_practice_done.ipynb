# Задача 1. Счётчик

Реализуйте класс <code style="color:red;background:none">Counter</code>, который работает как простой счётчик.

**Методы:**
* `__init__(start=0)` — создаёт счётчик с начальным значением `start` (по умолчанию 0),
* `increment()` — увеличивает значение на 1,
* `decrement()` — уменьшает значение на 1,
* `reset()` — сбрасывает значение обратно к начальному,
* `value()` — возвращает текущее значение.

```python
c = Counter(10)
c.increment()
c.increment()
print(c.value())   # 12
c.decrement()
print(c.value())   # 11
c.reset()
print(c.value())   # 10
```

````python
# Ваше решение
class Counter:

  def __init__(self,start=0):
    self.start = start
    self.counter = start
  def increment(self,):
    self.counter += 1
  def decrement(self,):
    self.counter -= 1

  def reset(self):
    self.counter = self.start

  def value(self):
    return self.counter


c = Counter(10)
c.increment()
c.increment()
print(c.value())   # 12
c.decrement()
print(c.value())   # 11
c.reset()
print(c.value())
````

# Задача 2. Стек

Реализуйте класс <code style="color:red;background:none">Stack</code> — структуру данных «стек» (последним пришёл — первым вышел, LIFO).

**Методы:**
* `push(item)` — добавить элемент на вершину стека,
* `pop()` — удалить и вернуть элемент с вершины стека. Если стек пуст — бросить `RuntimeError`,
* `peek()` — вернуть верхний элемент, не удаляя его. Если стек пуст — бросить `RuntimeError`,
* `is_empty()` — вернуть `True`, если стек пуст,
* `size()` — вернуть количество элементов в стеке.

```python
s = Stack()
s.push(1)
s.push(2)
s.push(3)
print(s.peek())    # 3
print(s.pop())     # 3
print(s.size())    # 2
print(s.is_empty()) # False
```

````python
class Stack:
  def __init__(self):
    self.stack = []

  def push(self, item):
    self.stack.append(item)

  def pop(self):
    if self.is_empty():
      raise RuntimeError(f"Stack is empty!")

    return self.stack.pop(-1)

  def peek(self):
    if self.is_empty():
      raise RuntimeError(f"Stack is empty!")

    return self.stack[-1]

  def size(self):
    return len(self.stack)

  def is_empty(self):
    # return len(self.stack) == 0
    return not bool(self.size())




s = Stack()
s.push(1)
s.push(2)
s.push(3)
print(s.peek())    # 3
print(s.pop())     # 3
print(s.size())    # 2
print(s.is_empty()) # False
````

````python
# Ваше решение
class Stack:
  def __init__(self):
    self.stack = []

  def push(self,x):
    self.stack.append(x)

  def pop(self):
    if len(self.stack) == 0:
      raise RuntimeError('Stack is empty')
    return self.stack.pop(-1)
  def peek(self):
    if len(self.stack) == 0:
      raise RuntimeError('Stack is empty')
    return self.stack[-1]

  def size(self):
    return len(self.stack)

  def is_empty(self):
    return bool(len(self.stack))
````

# Задача 3. Банковский счёт

Реализуйте класс <code style="color:red;background:none">BankAccount</code>, моделирующий банковский счёт.

**Методы:**
* `__init__(owner, balance=0)` — создать счёт с владельцем и начальным балансом,
* `deposit(amount)` — пополнить счёт на `amount`. Если сумма ≤ 0 — бросить `ValueError`,
* `withdraw(amount)` — снять `amount`. Если сумма ≤ 0 или на счёте недостаточно средств — бросить `ValueError`,
* `get_balance()` — вернуть текущий баланс,
* `get_history()` — вернуть список всех операций в формате `'+ 500'` или `'- 200'`.

```python
acc = BankAccount('Иван', 1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.get_balance())   # 1300
print(acc.get_history())   # ['+ 500', '- 200']
acc.withdraw(9999)         # ValueError
```

````python
# Ваше решение
class BankAccount:
  def __init__(self, owner, balance = 0):
    self.__owner = owner
    self.__balance = balance
    self.__history = []

  def deposit(self, amount):

    self.__balance += amount

    if self.__balance <= 0:
      raise ValueError('Отрицательный баланс')

    self.__history.append('+'+str(amount))


  def withdraw(self,amount):

    if self.__balance <= 0:
      raise ValueError('Отрицательный баланс')

    if self.__balance < amount:
      raise ValueError('Маленький баланс баланс')

    self.__balance -= amount
    self.__history.append('-'+str(amount))

  def get_balance(self):
    return self.__balance

  def get_history(self):
    return self.__history

acc = BankAccount('Иван', 1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.get_balance())   # 1300
print(acc.get_history())   # ['+ 500', '- 200']
acc.withdraw(9999)         # ValueError
````

# Задача 5. Матрица

Реализуйте класс <code style="color:red;background:none">Matrix</code> для работы с двумерными матрицами.

**Методы:**
* `__init__(rows, cols, fill=0)` — создать матрицу размером `rows × cols`, заполненную значением `fill`,
* `set(row, col, value)` — установить значение ячейки,
* `get(row, col)` — получить значение ячейки,
* `row_sum(row)` — вернуть сумму элементов строки,
* `col_sum(col)` — вернуть сумму элементов столбца,
* `display()` — красиво вывести матрицу в консоль.

```python
m = Matrix(3, 3)
m.set(0, 0, 1)
m.set(1, 1, 5)
m.set(2, 2, 9)
m.display()
# 1 0 0
# 0 5 0
# 0 0 9
print(m.row_sum(1))  # 5
print(m.col_sum(2))  # 9
```

````python
# Ваше решение

class Matrix:
  def __init__(self,rows, cols,fill=0):

    self.matrix = []
    for row in range(rows):
      self.matrix.append([])
      for col in range(cols):
          self.matrix[row].append(fill)

    self.rows = rows
    self.cols = cols
    self.fill = fill

  def set(self, row, col, value):
    self.matrix[row][col] = value

  def get(self,row,col):
    return self.matrix[row][col]

  def row_sum(self, row):
    return sum(self.matrix[row])

  def col_sum(self, col):
    s = 0
    for row in range(self.rows):
      s += self.matrix[row][col]
    return s

  def display(self):

    for row in self.matrix:
      for col in row:
        print(col, end = ' ')
      print()

m = Matrix(3, 3)
m.set(0, 0, 1)
m.set(1, 1, 5)
m.set(2, 2, 9)
m.display()
# 1 0 0
# 0 5 0
# 0 0 9
print(m.row_sum(1))  # 5
print(m.col_sum(2))  # 9
````
