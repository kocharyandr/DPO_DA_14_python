# Зачем нужны классы?

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Классы в Python — это способ работать с объектом, у которого есть **состояние**. Состояние — это данные, которые объект хранит внутри себя и которые могут меняться со временем.

Чтобы читать или изменять это состояние, используются **методы** — специальные функции, у которых есть доступ к содержимому объекта через параметр `self`.

Класс решает несколько важных задач:
* **Инкапсуляция** — пользователю не нужно знать, как именно хранятся данные внутри. Он просто вызывает методы.
* **Гибкость** — внутреннее устройство можно менять, не ломая код пользователей.
* **Повторное использование** — один класс можно создать и использовать в разных частях программы.
</div>

## Начальный класс: состояние номеров

Что является **состоянием отеля**? Для начала — только информация о том, какие номера заняты, а какие свободны.

Будем хранить это в списке длиной `num_of_rooms`:
* `0` — номер свободен,
* `1` — номер занят.

Индекс элемента в списке — это и есть номер комнаты. Например, `rooms[3] == 1` означает, что третий номер занят.

Добавим два базовых метода:
* <code style="color:red;background:none">occupy(room_id)</code> — заселить гостей в номер,
* <code style="color:red;background:none">free(room_id)</code> — освободить номер.

````python
class Hotel:
    """Класс для управления состоянием номеров в отеле."""

    def __init__(self, num_of_rooms):
        # Список из нулей: 0 — свободно, 1 — занято
        self.rooms = [0 for _ in range(num_of_rooms)]

    def occupy(self, room_id):
        """Заселить гостей в номер room_id."""
        self.rooms[room_id] = 1

    def free(self, room_id):
        """Освободить номер room_id."""
        self.rooms[room_id] = 0
````

Создадим отель на 10 номеров и попробуем поработать с ним:

````python
# Создаём объект — отель с 10 номерами
hotel = Hotel(num_of_rooms=10)
hotel.rooms  # все нули: все номера свободны
````

````python
# Гости заселяются в номера 3 и 6
hotel.occupy(3)
hotel.occupy(6)
hotel.rooms  # в позициях 3 и 6 теперь стоит 1
````

````python
# Гости из номера 3 уехали
hotel.free(3)
hotel.rooms  # позиция 3 снова 0
````

### А зачем класс? Разве нельзя просто написать функцию?

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Можно написать и так:

```python
def occupy(rooms, room_id):
    rooms[room_id] = 1
    return rooms
```

Но класс даёт важное преимущество: тем, кто пользуется классом (включая вас самих), **не нужно знать**, как внутри хранятся данные. Если завтра вы решите заменить `list` на `dict` (например, так быстрее), пользователи ничего не заметят — их код не изменится. То же самое с функциональностью: добавление нового метода не сломает существующий код. А это очень важно.

# Задание 1. Заполняемость и закрытие отеля

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Добавьте в класс `Hotel` два новых метода:

* <code style="color:red;background:none">occupy_rate()</code> — возвращает **долю занятых номеров** (от 0.0 до 1.0). Например, если занято 3 из 10 номеров, метод возвращает `0.3`.
* <code style="color:red;background:none">close()</code> — **освобождает все номера** (как будто отель закрылся на ремонт). После вызова `close()` метод `occupy_rate()` должен вернуть `0.0`.
</div>

**Подсказка.** `occupy_rate()` — это количество занятых номеров (сумма единиц в списке), делённое на общее количество номеров. Метод `close()` удобно реализовать через уже написанный метод `free()`, пройдясь по всем номерам в цикле.

````python
class Hotel:
    """Класс для управления состоянием номеров в отеле."""

    def __init__(self, num_of_rooms):
        self.rooms = [0 for _ in range(num_of_rooms)]
        self.num_of_rooms = num_of_rooms  # сохраняем отдельно, чтобы не считать каждый раз

    def occupy(self, room_id):
        """Заселить гостей в номер room_id."""
        self.rooms[room_id] = 1

    def free(self, room_id):
        """Освободить номер room_id."""
        self.rooms[room_id] = 0

    def occupy_rate(self):
        """Вернуть долю занятых номеров (от 0.0 до 1.0)."""
        return sum(self.rooms) / self.num_of_rooms

    def close(self):
        """Освободить все номера.
        После вызова occupy_rate() вернёт 0.0.
        """
        for room_id in range(self.num_of_rooms):
            self.free(room_id)
````

````python
hotel = Hotel(num_of_rooms=10)
hotel.occupy(3)
hotel.occupy(7)
hotel.rooms
````

````python
hotel.occupy_rate()  # 2 из 10 заняты → 0.2
````

````python
hotel.close()        # освобождаем все номера
hotel.occupy_rate()  # → 0.0
````

# Задание 2. Защита от ошибок

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Сейчас методы `occupy` и `free` не проверяют корректность действия: можно заселить гостей в уже занятый номер или освободить уже свободный. Нужно это исправить.

Доработайте методы `occupy` и `free`:
* `occupy(room_id)` — проверять, что номер свободен перед заселением. Если занят — бросать исключение.
* `free(room_id)` — проверять, что номер занят перед освобождением. Если уже свободен — бросать исключение.

Исключение бросается командой:
```python
raise RuntimeError("Текст сообщения об ошибке")
```
</div>

**Зачем это нужно?** Без проверок ошибки в логике программы останутся незаметными: данные молча испортятся, и найти причину будет трудно. Явное исключение с понятным сообщением сразу укажет на проблему.

````python
class Hotel:
    """Класс для управления состоянием номеров в отеле."""

    def __init__(self, num_of_rooms):
        self.rooms = [0 for _ in range(num_of_rooms)]
        self.num_of_rooms = num_of_rooms

    def occupy(self, room_id):
        """Заселить гостей в номер room_id. Если номер уже занят — ошибка."""
        if self.rooms[room_id] == 0:
            self.rooms[room_id] = 1
        else:
            raise RuntimeError(f"Номер {room_id} уже занят!")

    def free(self, room_id):
        """Освободить номер room_id. Если номер уже свободен — ошибка."""
        if self.rooms[room_id] == 1:
            self.rooms[room_id] = 0
        else:
            raise RuntimeError(f"Номер {room_id} уже свободен!")

    def occupy_rate(self):
        """Вернуть долю занятых номеров."""
        return sum(self.rooms) / self.num_of_rooms

    def close(self):
        """Освободить все номера."""
        for room_id in range(self.num_of_rooms):
            self.free(room_id)
````

> * `RuntimeError` с понятным сообщением информативнее, чем просто `AssertionError`.

# Задание 3. Бронирование номеров

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Добавьте возможность бронировать номера на конкретную дату:

* <code style="color:red;background:none">is_booked(date, room_id)</code> — возвращает `True`, если номер `room_id` уже забронирован на дату `date`, иначе `False`.
* <code style="color:red;background:none">book(date, room_id)</code> — бронирует номер `room_id` на дату `date`. Если номер на эту дату уже забронирован — бросить исключение.

Бронирования храним в словаре `booked_rooms`: ключ — дата (строка), значение — список забронированных номеров.
</div>

**Структура данных для бронирований:**

```python
booked_rooms = {
    '13.08.2025': [9, 8],
    '20.08.2025': [8],
}
```

Обратите внимание: бронирование на дату **не зависит** от текущего состояния номера (`self.rooms`). Гость может забронировать номер заранее, пока он ещё свободен сегодня.

````python
class Hotel:
    """Класс для управления состоянием номеров в отеле."""

    def __init__(self, num_of_rooms):
        self.rooms = [0 for _ in range(num_of_rooms)]
        self.num_of_rooms = num_of_rooms
        self.booked_rooms = {}  # словарь: дата → список забронированных номеров

    def occupy(self, room_id):
        """Заселить гостей в номер room_id."""
        if self.rooms[room_id] == 0:
            self.rooms[room_id] = 1
        else:
            raise RuntimeError(f"Номер {room_id} уже занят!")

    def free(self, room_id):
        """Освободить номер room_id."""
        if self.rooms[room_id] == 1:
            self.rooms[room_id] = 0
        else:
            raise RuntimeError(f"Номер {room_id} уже свободен!")

    def occupy_rate(self):
        """Вернуть долю занятых номеров."""
        return sum(self.rooms) / self.num_of_rooms

    def close(self):
        """Освободить все номера."""
        for room_id in range(self.num_of_rooms):
            self.free(room_id)

    def is_booked(self, date, room_id):
        """Проверить, забронирован ли номер room_id на дату date."""
        if date in self.booked_rooms:
            return room_id in self.booked_rooms[date]
        return False

    def book(self, date, room_id):
        """Забронировать номер room_id на дату date.
        Если номер уже забронирован на эту дату — бросить исключение.
        """
        if self.is_booked(date, room_id):
            raise RuntimeError(f"Номер {room_id} уже забронирован на {date}!")
        elif date in self.booked_rooms:
            self.booked_rooms[date].append(room_id)  # дата уже есть — добавляем номер
        else:
            self.booked_rooms[date] = [room_id]      # новая дата — создаём запись
````

````python
hotel = Hotel(num_of_rooms=10)
hotel.occupy(3)
hotel.occupy(7)
````

````python
# Бронируем номер 9 на 13 августа
hotel.book('13.08.2025', 9)
hotel.booked_rooms
````

````python
hotel.is_booked('13.08.2025', 9)  # True — забронирован
````

````python
hotel.is_booked('14.08.2025', 9)  # False — на другую дату брони нет
````

````python
# Попробуем забронировать номер 9 на 13 августа повторно — ошибка
try:
    hotel.book('13.08.2025', 9)
except RuntimeError as e:
    print('Ошибка:', e)
````

````python
# Бронируем номер 8 на ту же дату — ок
hotel.book('13.08.2025', 8)

# И номер 8 на другую дату — тоже ок (разные даты независимы)
hotel.book('20.08.2025', 8)

hotel.booked_rooms
````

# Задание 4. Выручка за день

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Добавьте метод <code style="color:red;background:none">income(date)</code>, который возвращает выручку отеля за указанную дату.

Для простоты будем считать, что стоимость **всех номеров одинакова** и составляет **200$**. Таким образом, выручка за день = количество забронированных номеров × 200.

Если на указанную дату нет ни одного бронирования, метод должен вернуть `0`, а не падать с ошибкой.
</div>

````python
class Hotel:
    """Класс для управления состоянием номеров в отеле."""

    ROOM_PRICE = 200  # стоимость номера — статический атрибут класса

    def __init__(self, num_of_rooms):
        self.rooms = [0 for _ in range(num_of_rooms)]
        self.num_of_rooms = num_of_rooms
        self.booked_rooms = {}

    def occupy(self, room_id):
        """Заселить гостей в номер room_id."""
        if self.rooms[room_id] == 0:
            self.rooms[room_id] = 1
        else:
            raise RuntimeError(f"Номер {room_id} уже занят!")

    def free(self, room_id):
        """Освободить номер room_id."""
        if self.rooms[room_id] == 1:
            self.rooms[room_id] = 0
        else:
            raise RuntimeError(f"Номер {room_id} уже свободен!")

    def occupy_rate(self):
        """Вернуть долю занятых номеров (от 0.0 до 1.0)."""
        return sum(self.rooms) / self.num_of_rooms

    def close(self):
        """Освободить все номера."""
        for room_id in range(self.num_of_rooms):
            self.free(room_id)

    def is_booked(self, date, room_id):
        """Проверить, забронирован ли номер room_id на дату date."""
        if date in self.booked_rooms:
            return room_id in self.booked_rooms[date]
        return False

    def book(self, date, room_id):
        """Забронировать номер room_id на дату date."""
        if self.is_booked(date, room_id):
            raise RuntimeError(f"Номер {room_id} уже забронирован на {date}!")
        elif date in self.booked_rooms:
            self.booked_rooms[date].append(room_id)
        else:
            self.booked_rooms[date] = [room_id]

    def income(self, date):
        """Вернуть выручку за дату date.
        Если на эту дату нет бронирований — возвращает 0.
        """
        booked_today = self.booked_rooms.get(date, [])  # [] если даты нет в словаре
        total = len(booked_today) * self.ROOM_PRICE
        print(f'Доход за {date}: {total}$')
        return total
````

> **Обратите внимание:** вместо `self.booked_rooms[date]` используем `self.booked_rooms.get(date, [])`. Метод `.get()` возвращает значение по ключу, а если ключа нет — возвращает второй аргумент (у нас `[]`). Это защищает от `KeyError` при запросе незабронированной даты.

````python
hotel = Hotel(num_of_rooms=10)
hotel.occupy(3)
hotel.occupy(7)

hotel.book('13.08.2025', 9)
hotel.book('13.08.2025', 8)
hotel.book('20.08.2025', 8)
````

````python
hotel.income('13.08.2025')  # 2 номера × 200$ = 400$
````

````python
hotel.income('20.08.2025')  # 1 номер × 200$ = 200$
````

````python
hotel.income('01.01.2025')  # нет бронирований → 0$
````

# Итоговый код класса

Ниже — полная версия класса `Hotel` со всеми методами, реализованными в заданиях 1–4.

| Метод | Описание |
|-------|----------|
| `__init__(num_of_rooms)` | Создать отель с заданным числом номеров |
| `occupy(room_id)` | Заселить гостей в номер (с проверкой) |
| `free(room_id)` | Освободить номер (с проверкой) |
| `occupy_rate()` | Доля занятых номеров |
| `close()` | Освободить все номера |
| `is_booked(date, room_id)` | Проверить бронирование номера на дату |
| `book(date, room_id)` | Забронировать номер на дату |
| `income(date)` | Выручка за указанную дату |

````python
class Hotel:
    """Система управления отелем.

    Позволяет отслеживать состояние номеров, принимать бронирования
    и рассчитывать выручку за произвольную дату.
    """

    ROOM_PRICE = 200  # стоимость одного номера в сутки, $

    def __init__(self, num_of_rooms):
        self.rooms = [0 for _ in range(num_of_rooms)]  # 0 — свободно, 1 — занято
        self.num_of_rooms = num_of_rooms
        self.booked_rooms = {}  # дата → [список забронированных номеров]

    def occupy(self, room_id):
        """Заселить гостей в номер room_id."""
        if self.rooms[room_id] == 0:
            self.rooms[room_id] = 1
        else:
            raise RuntimeError(f"Номер {room_id} уже занят!")

    def free(self, room_id):
        """Освободить номер room_id."""
        if self.rooms[room_id] == 1:
            self.rooms[room_id] = 0
        else:
            raise RuntimeError(f"Номер {room_id} уже свободен!")

    def occupy_rate(self):
        """Вернуть долю занятых номеров (от 0.0 до 1.0)."""
        return sum(self.rooms) / self.num_of_rooms

    def close(self):
        """Освободить все номера."""
        for room_id in range(self.num_of_rooms):
            self.free(room_id)

    def is_booked(self, date, room_id):
        """Проверить, забронирован ли номер room_id на дату date."""
        if date in self.booked_rooms:
            return room_id in self.booked_rooms[date]
        return False

    def book(self, date, room_id):
        """Забронировать номер room_id на дату date.
        Бросает RuntimeError, если номер уже забронирован на эту дату.
        """
        if self.is_booked(date, room_id):
            raise RuntimeError(f"Номер {room_id} уже забронирован на {date}!")
        elif date in self.booked_rooms:
            self.booked_rooms[date].append(room_id)
        else:
            self.booked_rooms[date] = [room_id]

    def income(self, date):
        """Вернуть выручку за дату date (количество бронирований × ROOM_PRICE)."""
        booked_today = self.booked_rooms.get(date, [])
        total = len(booked_today) * self.ROOM_PRICE
        print(f'Доход за {date}: {total}$')
        return total
````
