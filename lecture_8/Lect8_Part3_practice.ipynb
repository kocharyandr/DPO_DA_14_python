# Задача 1. Счётчик

Реализуйте класс <code style="color:red;background:none">Counter</code>, который работает как простой счётчик.

**Методы:**
* `__init__(start=0)` — создаёт счётчик с начальным значением `start` (по умолчанию 0),
* `increment()` — увеличивает значение на 1,
* `decrement()` — уменьшает значение на 1,
* `reset()` — сбрасывает значение обратно к начальному,
* `value()` — возвращает текущее значение.

```python
c = Counter(10)
c.increment()
c.increment()
print(c.value())   # 12
c.decrement()
print(c.value())   # 11
c.reset()
print(c.value())   # 10
```

# Задача 2. Стек

Реализуйте класс <code style="color:red;background:none">Stack</code> — структуру данных «стек» (последним пришёл — первым вышел, LIFO).

**Методы:**
* `push(item)` — добавить элемент на вершину стека,
* `pop()` — удалить и вернуть элемент с вершины стека. Если стек пуст — бросить `RuntimeError`,
* `peek()` — вернуть верхний элемент, не удаляя его. Если стек пуст — бросить `RuntimeError`,
* `is_empty()` — вернуть `True`, если стек пуст,
* `size()` — вернуть количество элементов в стеке.

```python
s = Stack()
s.push(1)
s.push(2)
s.push(3)
print(s.peek())    # 3
print(s.pop())     # 3
print(s.size())    # 2
print(s.is_empty()) # False
```

# Задача 7. Очередь с приоритетом

Реализуйте класс <code style="color:red;background:none">PriorityQueue</code> — очередь, в которой элементы извлекаются в порядке приоритета (чем меньше число приоритета, тем важнее задача).

**Методы:**
* `enqueue(item, priority)` — добавить элемент с заданным приоритетом,
* `dequeue()` — извлечь и вернуть элемент с наименьшим приоритетом. Если очередь пуста — бросить `RuntimeError`,
* `peek()` — посмотреть первый по приоритету элемент, не удаляя его,
* `size()` — вернуть количество элементов.

```python
pq = PriorityQueue()
pq.enqueue('Починить баг', 1)
pq.enqueue('Написать тесты', 3)
pq.enqueue('Сделать ревью', 2)
print(pq.dequeue())  # 'Починить баг'   (приоритет 1)
print(pq.dequeue())  # 'Сделать ревью'  (приоритет 2)
print(pq.size())     # 1
```

# Задача 10. Социальная сеть

Реализуйте класс <code style="color:red;background:none">SocialNetwork</code>, моделирующий упрощённую социальную сеть.

**Методы:**
* `register(username)` — зарегистрировать пользователя. Если имя уже занято — бросить `ValueError`,
* `follow(who, whom)` — пользователь `who` подписывается на `whom`. Если кто-то из них не зарегистрирован или подписка уже есть — бросить соответствующее исключение,
* `unfollow(who, whom)` — отписаться,
* `get_followers(username)` — вернуть список тех, кто подписан на `username`,
* `get_following(username)` — вернуть список тех, на кого подписан `username`,
* `mutual_friends(user1, user2)` — вернуть список пользователей, на которых подписаны оба.

```python
sn = SocialNetwork()
sn.register('alice')
sn.register('bob')
sn.register('carol')
sn.follow('alice', 'bob')
sn.follow('alice', 'carol')
sn.follow('bob', 'carol')
print(sn.get_followers('carol'))         # ['alice', 'bob']
print(sn.get_following('alice'))         # ['bob', 'carol']
print(sn.mutual_friends('alice', 'bob')) # ['carol']
```
