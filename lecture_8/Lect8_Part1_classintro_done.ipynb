# Объектно-ориентированное программирование

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

**Объектно-ориентированное программирование (ООП)** — это подход к разработке программ, при котором код организован вокруг **объектов**, а не функций и процедур.

Ключевые понятия ООП:

* **Класс** — шаблон (чертёж) для создания объектов. Описывает, какими свойствами и поведением будут обладать объекты.
* **Объект (экземпляр)** — конкретный представитель класса, созданный по этому шаблону.
* **Атрибут** — переменная, хранящая свойство объекта или класса.
* **Метод** — функция внутри класса, описывающая поведение объекта.

Четыре принципа ООП: **инкапсуляция**, **наследование**, **полиморфизм**, **абстракция**.
</div>

# Создание классов в Python

Объявление класса начинается с ключевого слова <code style="color:red;background:none">class</code>, за которым следует имя класса. По договорённости имена классов записываются в стиле **CamelCase** (каждое слово с заглавной буквы).

Минимально допустимый класс выглядит так:

````python
class Car:
    """
    Класс, описывающий автомобиль.
    """
    pass  # пустое тело — используем pass, чтобы класс был синтаксически корректным

class Bugatti:
    """
    Класс, описывающий автомобиль.
    """
    pass  # пустое тело — используем pass, чтобы класс был синтаксически корректным
````

````python
a = Car()
type(a)
````

````python
a = Bugatti()
````

````python
a = Bugatti()
````

Класс состоит из трёх частей:

* **Объявление** — ключевое слово <code style="color:red;background:none">class</code> и имя класса.
* **Строка документации (docstring)** — необязательна, но является хорошим тоном: сразу после объявления описывается назначение класса.
* **Тело класса** — атрибуты и методы.

Несмотря на пустое тело, на основе класса <code style="color:red;background:none">Car</code> уже можно создавать объекты — **экземпляры класса**:

````python
audi = Car()  # создаём экземпляр класса Car
print(type(audi))  # убеждаемся, что audi — это объект класса Car
````

## Статические и динамические атрибуты

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Атрибуты класса делятся на два вида:

* **Статический (атрибут класса)** — объявляется прямо в теле класса, общий для всех экземпляров. Доступен без создания объекта через имя класса: `Car.default_color`.
* **Динамический (атрибут экземпляра)** — объявляется через <code style="color:red;background:none">self</code> внутри методов, уникален для каждого объекта. Доступен только через экземпляр: `bmw.color`.
</div>

````python
class Car:
    default_color = "blue"  # статический атрибут — общий для всех машин

    def __init__(self, color, brand, doors_num):
        # Если цвет не задан явно — используем стандартный

        if color is None:

            self.color = self.default_color
        else:
            self.color = color

        self.brand = brand          # динамические атрибуты —
        self.doors_num = doors_num  # уникальны для каждого экземпляра
````

### Доступ к статическому атрибуту

Статический атрибут <code style="color:red;background:none">default_color</code> доступен через имя класса — без создания объекта:

````python
Car.default_color  # не нужен ни один объект класса Car
````

А вот динамические атрибуты <code style="color:red;background:none">color</code>, <code style="color:red;background:none">brand</code> и <code style="color:red;background:none">doors_num</code> объявлены через <code style="color:red;background:none">self</code> — они привязаны к конкретному объекту и требуют его создания:

````python
bmw = Car(None, "BMW", 2)
print(bmw.brand)     # BMW
print(bmw.color)     # blue — цвет не был задан, применился стандартный
print(bmw.doors_num) # 2
````

Попытка обратиться к динамическому атрибуту через имя класса (без объекта) вызовет ошибку <code style="color:red;background:none">AttributeError</code>:

````python
Car.brand  # AttributeError: type object 'Car' has no attribute 'brand'
````

### Изменение статического атрибута

Если изменить статический атрибут **через имя класса**, изменение отразится на всех экземплярах, у которых нет одноимённого атрибута экземпляра:

````python
# int == Car
# a = int('10') == bmw
````

````python
bmw.default_color
````

````python
Car.default_color = "red"
print(Car.default_color)
````

````python
bmw.default_color
````

Создадим два объекта и убедимся, что у обоих <code style="color:red;background:none">default_color</code> одинаковый:

````python
bmw = Car(None, "BMW", 2)
audi = Car(None, "AUDI", 4)

print(bmw.default_color)   # red
print(audi.default_color)  # red
````

### Важный нюанс: перекрытие статического атрибута

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Если изменить <code style="color:red;background:none">default_color</code> **через экземпляр**, Python не изменит статический атрибут — вместо этого он создаст у этого экземпляра **новый атрибут экземпляра** с таким же именем. Статический атрибут класса при этом остаётся нетронутым, но перекрывается локальным.
</div>

````python
bmw.default_color = "blue"  # создаётся атрибут ЭКЗЕМПЛЯРА, не меняется атрибут КЛАССА

print(bmw.default_color)   # blue  — смотрит на свой атрибут экземпляра
print(audi.default_color)  # red   — у audi своего атрибута нет, читает из класса
print(Car.default_color)   # red   — атрибут класса не изменился
````

### Аналогия с автозаводом

Представьте, что класс `Car` — это **автозавод**. Завод выпускает все машины в одном стандартном цвете (`default_color`). Если покупатель хочет другой цвет — он указывает его явно при заказе. Если завод решает сменить стандартный цвет — все **новые** машины будут выходить уже в новом цвете.

````python
# Завод выпускает машины в зелёном цвете
Car.default_color = "green"

car1 = Car(None, "Niva", 2)
car2 = Car(None, "Niva", 2)
car3 = Car(None, "Niva", 4)
car4 = Car("black", "Niva", 4)  # этот покупатель выбрал чёрный


print(car1.color, car2.color, car3.color, car4.color)

# Завод переходит на красный цвет
Car.default_color = "red"

car5 = Car(None, "Niva", 2)
car6 = Car("olive", "Niva", 2)  # этот выбрал оливковый
car7 = Car(None, "Niva", 4)
car8 = Car(None, "Niva", 4)

# Машины из первой партии не изменились — их цвет был задан при создании
print(car1.color, car2.color, car3.color, car4.color)
# Новые машины — уже красные (или с индивидуальным цветом)
print(car5.color, car6.color, car7.color, car8.color)
````

## Конструктор <code style="color:red;background:none">__init__</code> и аргумент <code style="color:red;background:none">self</code>

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

**<code style="color:red;background:none">__init__</code>** — специальный метод-конструктор, который Python вызывает автоматически при создании каждого нового объекта. Именно здесь задаются начальные значения атрибутов экземпляра.

**<code style="color:red;background:none">self</code>** — ссылка на конкретный экземпляр класса, через который вызывается метод. Он нужен, чтобы разграничить атрибуты разных объектов: `bmw.color` и `audi.color` — это два разных атрибута двух разных объектов, и `self` указывает, к какому именно обращаться.

<code style="color:red;background:none">self</code> должен быть **первым параметром** в каждом методе экземпляра. При вызове метода Python передаёт его автоматически — вам не нужно указывать его явно.
</div>

````python
class Car:
    default_color = "green"

    def __init__(self, color, brand, doors_num):
        # self.color — атрибут этого конкретного объекта
        if color is None:
            self.color = self.default_color  # self. — обращаемся к атрибуту ЭТОГО объекта
        else:
            self.color = color

        self.brand = brand
        self.doors_num = doors_num

fiat = Car(None, "Fiat", 5)
print(fiat.color)  # green
````

### Что произойдёт без <code style="color:red;background:none">self</code>?

Если написать просто `default_color` вместо `self.default_color`, Python будет искать переменную в локальной области видимости функции — и не найдёт её, так как статический атрибут принадлежит классу, а не локальной области:

````python
class Car:
    default_color = "green"

    def __init__(self, color, brand, doors_num):
        if color is None:
            self.color = default_color  # NameError! Python ищет локальную переменную
        else:
            self.color = color

        self.brand = brand
        self.doors_num = doors_num

fiat = Car(None, "Fiat", 5)  # ← здесь будет ошибка
````

## Методы класса

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

**Метод** — это функция, определённая внутри класса. В Python методы делятся на три вида:

| Вид метода | Декоратор | Первый параметр | К чему имеет доступ |
|------------|-----------|-----------------|---------------------|
| Метод экземпляра | нет | `self` | объект и класс |
| Метод класса | `@classmethod` | `cls` | только класс |
| Статический метод | `@staticmethod` | нет | ничего (только аргументы) |
</div>

````python
class Car:

    @staticmethod
    def ex_static_method():
        print("Это статический метод")

    @classmethod
    def ex_class_method(cls):
        print("Это метод класса, cls =", cls)

    def ex_method(self):
        print("Это метод экземпляра, self =", self)
````

Статический и классовый методы можно вызвать **напрямую через класс**, не создавая объект. Метод экземпляра так вызвать нельзя:

````python
Car.ex_static_method()   # ок
Car.ex_class_method()    # ок
Car.ex_method()          # TypeError: ex_method() missing 1 required positional argument: 'self'
````

````python
bwm = Car()
bwm.ex_method()  # ок — Python автоматически передаёт self
````

````python
bwm.ex_static_method()
bwm.ex_class_method()
````

### Статический метод (<code style="color:red;background:none">@staticmethod</code>)

Статическому методу **не нужны** ни `self`, ни `cls`. Он работает только с теми данными, которые явно переданы в него как аргументы. По сути это обычная функция, «прикреплённая» к классу для логической группировки.

Статические методы **не могут** ни читать, ни изменять состояние класса или его объектов. Типичное применение — вспомогательные вычисления, валидация данных.

### Метод класса (<code style="color:red;background:none">@classmethod</code>)

Метод класса получает **сам класс** в виде первого аргумента `cls`. Он может читать и изменять атрибуты класса, что отразится на всех объектах. Но он не имеет доступа к конкретному экземпляру.

Наиболее частое применение — **альтернативные конструкторы**: создание объекта из данных в нестандартном формате. Встроенный пример — <code style="color:red;background:none">dict.fromkeys()</code>:

````python
dict.fromkeys('AEIOU')  # вызывается на классе dict, возвращает новый словарь
````

### Метод экземпляра

Самый распространённый вид. Принимает объект класса как первый аргумент (`self`) и может читать/изменять как атрибуты экземпляра, так и атрибуты класса через `self.__class__`.

Встроенный пример — <code style="color:red;background:none">str.upper()</code>:

````python
"welcome".upper()  # вызывается на конкретной строке — экземпляре класса str
````

````python
str.isalpha('a')
````

````python
"welcome".isalpha()
````

### Практический пример: все три вида методов вместе

Рассмотрим класс `Car`, в котором все три вида методов выполняют осмысленные роли:

````python
from datetime import date

class Car:
    """ Класс отражающие машины в общности """
    def __init__(self, brand, age):
        self.brand = brand
        self.age = age

    @classmethod
    def from_production_year(cls, brand, prod_year):
        """Альтернативный конструктор: создаёт объект по году выпуска."""
        # return cls.__init__(brand, date.today().year - prod_year) - условно

        return cls(brand, date.today().year - prod_year)

    @staticmethod
    def is_warranty_active(age):
        """Проверяет, действует ли гарантия (первые 3 года)."""
        return age < 3

    def info(self):
        """Метод экземпляра: выводит информацию об автомобиле."""
        print('Car: ' + self.brand)
        print('Age: ' + str(self.age))
        if self.is_warranty_active(self.age):
            print('Гарантия: АКТИВНА')
        else:
            print('Гарантия: истекла')

car1 = Car('Subaru', 5)
car2 = Car.from_production_year('Skoda', 2024)  # альтернативный конструктор
````

````python
car1.brand, car1.age
````

````python
car2.brand, car2.age
````

````python
Car.is_warranty_active(25)  # статический метод — вызываем прямо на классе
````

````python
car1.info()
````

````python
car2.info()
````

Подводим итог по выбору типа метода:

* **Метод экземпляра** — нужен доступ к конкретному объекту (`self`). Используется чаще всего.
* **Метод класса** — нужен альтернативный конструктор или операция над всем классом (`cls`). Пример: `from_production_year` — создаёт объект по году выпуска.
* **Статический метод** — вспомогательная функция, не зависящая от состояния объекта или класса. Пример: `is_warranty_active` — просто проверяет число.

## Инкапсуляция и уровни доступа

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

**Инкапсуляция** — принцип ООП, предполагающий сокрытие внутреннего устройства объекта и предоставление доступа к данным только через специальный интерфейс.

В Java, C++ и C# доступ к атрибутам контролируется ключевыми словами (`private`, `protected`, `public`). В Python такого механизма нет, но есть **соглашение**:

| Префикс | Смысл | Пример |
|---------|-------|--------|
| без префикса | публичный | `self.brand` |
| одно подчёркивание `_` | «защищённый» (по договорённости) | `self._brand` |
| два подчёркивания `__` | «приватный» (name mangling) | `self.__brand` |

Для чтения и изменения атрибутов используют методы **getter** и **setter**.
</div>

### Одно подчёркивание: «не трогай снаружи»

Одно подчёркивание — это **соглашение**: атрибут считается внутренним, снаружи лучше его не трогать. Но технически это возможно — Python не запретит.

````python
class Car:
    def __init__(self, brand, doors_num):
        self._brand = brand          # «защищённые» атрибуты
        self._doors_num = doors_num

    def get_brand(self):         # getter — читаем атрибут через метод
        return self._brand

    def set_brand(self, b):      # setter — меняем атрибут через метод
        self._brand = b

    def get_doors(self):
        return self._doors_num

    def set_doors(self, d):
        self._doors_num = d

    def info(self):
        return 'Nice car with ' + str(self._doors_num) + ' doors'
````

````python
mersedes = Car("Mercedes", 6)
print(mersedes.get_brand())  # правильно — через getter
print(mersedes._brand)       # работает, но нарушает соглашение
````

````python
mersedes._doors_num
````

### Два подчёркивания: name mangling

Если атрибут начинается с **двух подчёркиваний**, Python применяет механизм **name mangling** (искажение имени): атрибут `__brand` в классе `Car` на самом деле хранится под именем `_Car__brand`. Это затрудняет случайный доступ снаружи, но не делает атрибут по-настоящему приватным.

````python
class Car:
    def __init__(self, brand, doors_num):
        self.__brand = brand         # «приватные» атрибуты
        self.__doors_num = doors_num

    def get_brand(self):
        return self.__brand

    def set_brand(self, b):
        self.__brand = b

    def get_doors(self):
        return self.__doors_num

    def set_doors(self, d):
        self.__doors_num = d

    def info(self):
        return 'Nice car with ' + str(self.__doors_num) + ' doors'
````

Попытка обратиться к `__brand` напрямую вызовет <code style="color:red;background:none">AttributeError</code> — нужно использовать getter:

````python
mersedes = Car("Mercedes", 6)
print(mersedes.get_brand())  # Mercedes — через getter
mersedes.__brand             # AttributeError!
````

Но через «искажённое» имя `_Car__brand` доступ всё же возможен — Python не создаёт настоящего запрета:

````python
mersedes._Car__brand  # 'Mercedes' — обходной путь, но лучше так не делать
````

## Наследование

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

**Наследование** позволяет создать новый класс на основе существующего, переняв его атрибуты и методы. Это избавляет от необходимости переписывать один и тот же код.

Правило: **класс-наследник** должен представлять собой более частный случай **класса-родителя**.

Синтаксис: имя родительского класса указывается в круглых скобках при объявлении наследника.

```python
class Наследник(Родитель):
    ...
```
</div>

````python
class Car:
    def __init__(self, brand, doors_num):
        self.__brand = brand
        self.__doors_num = doors_num

    def get_brand(self):
        return self.__brand

    def set_brand(self, b):
        self.__brand = b

    def get_doors(self):
        return self.__doors_num

    def set_doors(self, d):
        self.__doors_num = d

    def info(self):
        return 'Nice car with ' + str(self.__doors_num) + ' doors'
````

````python
class Truck(Car):  # Truck наследует Car

    def __init__(self, brand, doors_num, load_weight, axes):
        super().__init__(brand, doors_num)  # вызываем конструктор родителя
        self.__load_weight = load_weight    # добавляем свои атрибуты
        self.__axes = axes

    def get_load(self):
        return self.__load_weight

    def set_load(self, l):
        self.__load_weight = l

    def get_axes(self):
        return self.__axes

    def set_axes(self, a):
        self.__axes = a
````

<code style="color:red;background:none">super()</code> — ключевое слово для обращения к родительскому классу. Вызов `super().__init__(brand, doors_num)` запускает конструктор класса `Car`, инициализируя унаследованные атрибуты. После этого можно добавлять атрибуты, специфичные для грузовика.

Проверим, что унаследованные методы работают:

````python
truck = Truck("Kamaz", 2, 13000, 6)

print(truck.get_brand())  # метод унаследован от Car
print(truck.get_load())   # собственный метод Truck
````

````python
truck.set_axes(8)
print(truck.get_axes())
````

### Множественное наследование

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

Python поддерживает **множественное наследование** — класс может наследовать сразу несколько родительских классов и получать их методы и атрибуты.

```python
class Наследник(Родитель1, Родитель2):
    ...
```

Если у нескольких родителей есть методы с одинаковым именем, Python использует порядок разрешения метода (MRO — Method Resolution Order): сначала ищет в первом родителе, затем во втором и т. д.
</div>

````python
class Creature:
    def __init__(self, name):
        self.name = name

class Animal(Creature):
    def sleep(self):
        print(self.name + ' is sleeping')
    # def play(self):
    #     print(self.name + ' is playing____')

class Pet(Creature):
    def play(self):
        print(self.name + ' is playing')

class Dog(Animal, Pet):  # Dog наследует и Animal, и Pet
    def bark(self):
        print(self.name + ' is barking')

beast = Dog("Buddy")
beast.sleep()  # унаследован от Animal
beast.play()   # унаследован от Pet
beast.bark()   # собственный метод Dog
````

Объект класса `Dog` получает `name` из `Creature` (через оба родителя), методы `sleep` и `play` от родителей и добавляет собственный метод `bark`. Поскольку конструкторы в `Animal` и `Pet` не переопределены, при создании `Dog` Python автоматически вызовет конструктор `Creature.__init__`.

````python
class Dog(Creature, Animal):   # Creature — родитель Animal, но стоит первым
    pass
````

````python
class A(Animal, Pet): pass     # здесь Animal раньше Pet
class B(Pet, Animal): pass     # а здесь Pet раньше Animal

class C(A, B):                 # C должен соблюсти оба порядка
    pass
````

## Полиморфизм

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%">

**Полиморфизм** — возможность использовать один и тот же интерфейс для объектов разных классов. В контексте наследования это означает **переопределение метода** родительского класса в классе-наследнике.

Класс-наследник может полностью заменить реализацию метода родителя или расширить её, добавив специфическое поведение — при этом имя метода остаётся тем же.
</div>

В классе `Car` есть метод `info()`. Переопределим его в классе `Truck` так, чтобы он дополнительно выводил грузоподъёмность:

````python
class Car:
    def __init__(self, brand, doors_num):
        self.__brand = brand
        self.__doors_num = doors_num

    def get_brand(self):
        return self.__brand

    def set_brand(self, b):
        self.__brand = b

    def get_doors(self):
        return self.__doors_num

    def set_doors(self, d):
        self.__doors_num = d

    def info(self):
        return 'Nice car with ' + str(self.__doors_num) + ' doors'
````

````python
class Truck(Car):

    def __init__(self, brand, doors_num, load_weight, axes):
        super().__init__(brand, doors_num)
        self.__load_weight = load_weight
        self.__axes = axes

    def get_load(self):
        return self.__load_weight

    def set_load(self, l):
        self.__load_weight = l

    def get_axes(self):
        return self.__axes

    def set_axes(self, a):
        self.__axes = a

    def info(self):  # переопределяем метод родителя
        return ('Nice car with ' + str(self.get_doors()) +
                ' doors and can carry ' + str(self.__load_weight) + ' kg of cargo')
````

Один и тот же метод `info()` ведёт себя по-разному в зависимости от типа объекта:

````python
audi = Car("Audi", 4)
print(audi.info())   # версия из Car
````

````python
scania = Truck("Scania", 2, 6500, 4)
print(scania.info())  # переопределённая версия из Truck
````

**Итоговая схема принципов ООП, рассмотренных в этом занятии:**

| Принцип | Что означает | Где видели |
|---------|--------------|------------|
| **Инкапсуляция** | Скрываем внутреннее устройство, даём доступ через методы | `_brand`, `__brand`, getter/setter |
| **Наследование** | Класс-наследник перенимает свойства родителя | `Truck(Car)`, `Dog(Animal, Pet)` |
| **Полиморфизм** | Один метод — разное поведение в разных классах | `info()` в `Car` и `Truck` |
</div>

````python
class figure:
  def __init__(self, a, b):
    self.a = a
    self.b = b

  def square(self):
    return self.a * self.b


class triangle(figure):
  def __init__(self, a,b):
    super().__init__(a,b)

  def square(self):
    return self.a * self.b/2


class square(figure):
  def __init__(self,a):
    super().__init__(a,a)

````

````python
tr = triangle(2,2)
sq = triangle(2,2)

print(tr.square())
print(sq.square())
````
