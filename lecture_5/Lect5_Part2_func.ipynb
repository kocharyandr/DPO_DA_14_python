### Что такое функция в Python?

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;">
<b>Функция</b> — это именованный, переиспользуемый блок кода, предназначенный для выполнения конкретной задачи. Она принимает входные данные (аргументы), обрабатывает их и возвращает результат. Функции помогают избегать дублирования кода, делают программы структурированными и удобными для чтения. </div>

### Объявляя функцию, нужно следовать определенным правилам:

Объявление происходит с помощью ключевого слова <code style="color:red;background:none">def</code>, за ним идёт имя функции и круглые скобки <code style="color:red;background:none">()</code>.​

<code style="color:red;background:none">Аргументы</code>, передаваемые в функцию, должны находиться в круглых скобках. Там же можно определить их значения по умолчанию, указав их после знака равно.
​

Перед основным содержимым желательно включить <code style="color:red;background:none">строку документации</code> (docstring), которая обычно описывает назначение и основные принципы работы функции.
​

<code style="color:red;background:none">Тело функции</code> начинается после знака двоеточия, важно не забыть об отступах.
​

Чтобы выйти из функции в Python, используют оператор <code style="color:red;background:none">return [значение]</code>. Если оператор опущен, будет возвращено значение <code style="color:red;background:none">None</code>.

```python
def Имя(аргументы):
    """
    Документация
    """
    # Тело (инструкции)
    return значение
```

На самом деле, мы уже использовали огромное количество функций и методов: <code style="color:red;background:none">str()</code>, <code style="color:red;background:none">float()</code>, <code style="color:red;background:none">.add()</code>, <code style="color:red;background:none">.count()</code> и так далее.

## Простейшие функции и return

Попробуем самостоятельно реализовать функцию, складывающую два числа и выводящую результат на печать:

```python
def summarize(a, b):
    summ = a + b
    print(summ)
```

```python
x = summarize(1, 2)
print(x)
```

Правильнее написать с <code style="color:red;background:none">return</code>.

```python
def summarize(a, b):
    summ = a + b
    return summ
```

```python
x = summarize(1, 2)
print(x)
```

### Аргументы и значения по умолчанию

При описании функции в Python можно задать аргументы с начальными значениями — такие аргументы являются «необязательными».
​
* Сначала описывают обязательные параметры.
* После них – параметры со значениями по умолчанию.
* При вызове функции не обязательно указывать значения необязательных параметров.
* Если нужно изменить значение одного аргумента, не меняя другие, можно передать его по имени.
​

```python
def example(first, second=3, third=5):
    print(first)
    print(second)
    print(third)
    
example('my string')
```

```python
example('my string',second='dfgh')
```

А что, если параметров у функции несколько, но только мы не знаем, сколько?. Например, мы хотим складывать числа, которые пользователь вводит с клавиатуры, но мы заранее не знаем, сколько чисел он собирается ввести. Уже несколько раз мы упоминали такое понятие, как "распаковка" списков, кортежей и словарей (когда записывается что-то в духе \*list_name или \*\*dictionary_name). 

### Распаковка последовательностей и словарей

Если мы хотим содержимое двух списков положить в третий список или содержимое двух словарей положить в третий словарь, на помощь приходит «звёздочный синтаксис».

```python
list1 = [1,2,3]
list2 = [4,5,6]

list3 = [*list1,*list2] ##положили все элементы обоих списков в третий
list3
```

```python
dict1 = {'a':1,'b':2,'c':3}
dict2 = {'a':4,'e':5,'f':6}

dict3 = {**dict1,**dict2} ##положили все элементы обоих словарей в третий
dict3
```

Точно таким же образом можно что-то указать "в явном виде", а что-то распаковать из структуры (например, вы хотите дописать в новый список числа 3 и 4):

```python
list1 = [0,1,2]
print([3,4,list1]) # тут элементом стал сам список
print([3,4,*list1]) # а тут мы элементы списка распаковали
```

### *args и **kwargs

В Python можно передать большое количество аргументов двумя способами:​

<code style="color:red;background:none">*args</code> — для неименованных аргументов

<code style="color:red;background:none">**kwargs</code> — для именованных аргументов

Мы используем <code style="color:red;background:none">*args</code> и <code style="color:red;background:none">**kwargs</code>, когда заранее неизвестно, сколько значений хотим передать функции.

### \*args

<code style="color:red;background:none">\*args</code> нужен, когда мы хотим передать неизвестное количество позиционных (неименованных) аргументов.
​
Если поставить <code style="color:red;background:none">*</code> перед именем параметра, это имя будет принимать не один аргумент, а несколько; внутри функции они будут доступны как кортеж.

```python
def adder(*nums):
    summ = 0
    for n in nums:
        summ += n

    print("Sum: ", summ)

adder(3, 5)
adder(4, 5, 6, 7)
adder(1, 2, 3, 5, 6)
```

### **kwargs

<code style="color:red;background:none">**kwargs</code> используется для передачи переменного количества именованных аргументов.

Аргументы собираются в словарь и доступны под именем параметра без <code style="color:red;background:none">**</code>.
​

```python
def intro(**data):
    print("\nТип данных: ",type(data))

    for key, value in data.items():
        print(f"{key} - это {value}")

intro(Firstname="Иван", Lastname="Иванов", Age=50, Phone=1234567890)
intro(Firstname="Инна", Lastname="Иванова", Email="test@test.ru", Age=99, Phone=987654321)
```

В этом случае у нас есть функция **intro()** с параметром \*\*data. В функцию мы передали два словаря разной длины. Затем внутри функции мы прошлись в цикле по словарям, чтобы вывести их содержимое.

### Lambda-функции

Lambda-функции — это особый вид функций, которые объявляются с помощью ключевого слова <code style="color:red;background:none">lambda</code> вместо <code style="color:red;background:none">def</code>.​

* Принимают любое количество аргументов.
* Содержат только одно выражение.
* Всегда возвращают одно значение.

Общий синтаксис:

```python
lambda аргументы: выражение
```

```python
add_1 = lambda x: [i for i in range(x + 1)]
add_1(8) 
```

А если мы вдруг решили сложить два числа, то это тоже можно сделать с помощью лямбда-функции:

```python
add_2 = lambda x, y: x + y
add_2(3, 4)
```

Заработало! А если по-другому назвать аргументы?

```python
add_2 = lambda f123, er45: f123 + er45
add_2(3, 4)
```

Во всех трех примерах мы все-таки присвоили имя каждой из функций. Давайте попробуем вызвать функцию по-настоящему анонимно. Аргументы передаются в скобках после скобок, содержащих определение lambda функции

```python
(lambda x, y: x * y)(3, 5)
```

По сути, лямбда-функции умеют все, что и обычные функции, только они обязаны возвращать всего одно значение. 
* Они умеют работать с разными типами данных (Строками, например)
* Можно вызывать функцию без параметров
* Параметрам функции можно задать значения по умолчанию

```python
# Со строками
(lambda x, y: x * y)("Ха-",3)
```

```python
(lambda x, y: x + y)("Первая","Вторая")
```

```python
# А вот тут без аргументов
(lambda: [0,1,2,3])()
```

```python
# А тут есть значения по умолчанию
(lambda x=3, y = 5: str(x) + str(y))()
```

```python
# Здесь в качестве первого аргумента пришел список, а второй использовался по умолчанию
# На выходе должны получить первый элемент первого списка + 3
(lambda x,y=3: x[0] + y)([1,2,3])
```

```python
print((lambda x, y, z: x + y + z)(1, 2, 3)) # Три аргумента

print((lambda x, y, z=3: x + y + z)(1, 2)) # Три аргумента и у одного default-значение

print((lambda x, y, z=3: x + y + z)(1, y=2)) # Три аргумента, у одного default-значение и один мы передали "по имени"

print((lambda *args: sum(args))(1,2,3)) # Передали кортеж аргументов и сложили

print((lambda **kwargs: sum(kwargs.values()))(one=1, two=2, three=3)) # Передали словарь аргументов и сложили
```

### Функции map и filter с lambda

Lambda-функции часто применяются для преобразования коллекций через функции <code style="color:red;background:none">map</code> и <code style="color:red;background:none">filter</code>.

```python
a = [1,3,4,6]

def to_str(x):
    return str(x)

list(map(to_str, a))
```

```python
list(map(lambda x: str(x),a))
```

А еще есть прекрасная функция для фильтрации списков, которая работает аналогично map. Называется она filter. Пусть мы хотим отфильтровать все числа меньше 4 из списка. С помощью цикла это можно сделать вот так:

```python
numbers = [1,2,3,4,5]
numbers_under_4 = []
for number in numbers:
    if number < 4:
        numbers_under_4.append(number)
numbers_under_4
```

А с помощью спискового включения вот так:

```python
numbers = [1,2,3,4,5]
numbers_under_4 = [number for number in numbers if number < 4]
numbers_under_4
```

А чудо-функция filter сработает вот так:

```python
numbers = [1,2,3,4,5]
numbers_under_4 = list(filter(lambda x: x < 4, numbers))
numbers_under_4
```

Видно, что синтаксис такой же, как и у map. первый аргумент - функция, которую надо применять, а второй - коллекция, на которую надо применять. Но как и для map, функция возвращает итератор, поэтому, чтобы получить список, его нужно обернуть функцией list.

### Возврат нескольких значений

Нужно упомянуть еще один факт: по команде return функция может возвращать несколько значений. Например вот функция, которая умеет выводить сумму и разность двух аргументов функции:

```python
def sum_diff(a,b):
    return a+b, a-b

sum1, diff = sum_diff(4,1)
print(sum1)
print(diff)
```

### Глобальные и локальные переменные при объявлении функций

**Глобальные переменные** объявляются вне всех функций и доступны в любом месте программы — как внутри, так и вне функций.

```python
# Глобальная переменная
x = 10

def show_global():
    print("Глобальная переменная x внутри функции:", x)

show_global()

print("Глобальная переменная x вне функции:", x)
```

**Локальные переменные** объявляются внутри функции и доступны только в этой функции. За её пределами они не видны и не существуют.

```python
def show_local():
    y = 5  # локальная переменная
    print("Локальная переменная y:", y)

show_local()

print("Пытаемся обратиться к y вне функции:")
print(y)  # Ошибка: y не определена
```

**Одинаковые имена переменных** \
Если переменная с одинаковым именем определена и глобально, и локально, приоритет внутри функции имеет локальная переменная.

```python
x = 5  # глобальная

def foo():
    x = 10  # локальная
    print("Локальная x:", x)

foo()

print("Глобальная x:", x)
```

**Изменение глобальной переменной внутри функции** \
Для изменения глобальной переменной внутри функции нужно использовать ключевое слово global.

```python
count = 0
print("Значение до попадания в функцию:", count)

def increment():
    global count
    count += 1

increment()

print("Значение после попадания в функцию:", count)
```
