<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Регулярное выражение (regex)</b> — это специальный «мини-язык» для описания шаблонов строк. Оно задаёт правила, по которым мы находим, извлекаем или изменяем фрагменты текста. </div>

Регулярные выражения особенно полезны, когда:
* нужно найти все email-адреса в тексте;
* валидировать формат телефона или даты;
* извлечь нужные части строки (например, домен сайта);
* заменить одни шаблоны текста на другие.

В Python за работу с регулярными выражениями отвечает модуль <code style="color:red; background:none">re</code>.

## Ключевые функции модуля <code style="color:red; background:none">re</code>:

<code style="color:red; background:none">re.search(pattern, string)</code> — ищет первое совпадение где угодно в строке.

<code style="color:red; background:none">re.match(pattern, string)</code> — пытается сопоставить шаблон с начала строки.
​

<code style="color:red; background:none">re.fullmatch(pattern, string)</code> — проверяет, совпадает ли вся строка с шаблоном.
​

<code style="color:red; background:none">re.findall(pattern, string)</code> — возвращает список всех совпадений.
​

<code style="color:red; background:none">re.split(pattern, string)</code> — разбивает строку по шаблону (аналог <code style="color:red; background:none">str.split</code>, но по regex).

<code style="color:red; background:none">re.sub(pattern, repl, string)</code> — заменяет все совпадения шаблона на подстановку.

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Практический совет.</b> Начните с <code style="color:red; background:none">re.search</code>, <code style="color:red; background:none">re.findall</code> и <code style="color:red; background:none">re.sub</code>. Этого уже достаточно для 90% задач поиска и замены текста. </div>

## Символьные классы и метасимволы

**Базовые символьные классы:**

<code style="color:red; background:none">.</code> — любой символ, кроме переноса строки.
​

<code style="color:red; background:none">\d</code> — цифра (0–9).
​

<code style="color:red; background:none">\D</code> — «не цифра».
​

<code style="color:red; background:none">\w</code> — «слово» (буква, цифра, подчёркивание).
​

<code style="color:red; background:none">\W</code> — «не слово».
​

<code style="color:red; background:none">\s</code> — пробельный символ (пробел, табуляция, перенос строки).
​

<code style="color:red; background:none">\S</code> — «не пробельный».
​

**Квадратные скобки задают собственные наборы символов:**

<code style="color:red; background:none">[abc]</code> — один символ: <code style="color:red; background:none">a</code>, <code style="color:red; background:none">b</code> или <code style="color:red; background:none">c</code>.

<code style="color:red; background:none">[a-z]</code> — любая строчная латинская буква.

<code style="color:red; background:none">[0-9]</code> — любая цифра.

<code style="color:red; background:none">[^0-9]</code> — любой символ, кроме цифры.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Важно.</b> Выражение <code style="color:red; background:none">[0-9]</code> эквивалентно <code style="color:red; background:none">\d</code>. Внутри скобок большинство метасимволов теряют особый смысл, но <code style="color:red; background:none">^</code> в начале означает отрицание набора. </div>

```python
import re
```

```python
text = "ABC def 123_45\t\n"

print("Все цифры:", re.findall(r"\d", text))
```

```python
print("Все не-цифры:", re.findall(r"\D", text))
```

```python
print("Все 'словесные' символы:", re.findall(r"\w", text))
```

```python
print("Все пробелы/переносы строк:", re.findall(r"\s", text))
```

```python
text2 = "Почта: a1@b2.com, c3@d4.org-"
print("Буквы a или цифры 1-3:", re.findall(r"[a1-3]", text2))
print("Все символы КРОМЕ цифр:", re.findall(r"[^0-9]", text2))
```

В регулярных выражениях много обратных слэшей: <code style="color:red; background:none">\d</code>, <code style="color:red; background:none">\s</code>, <code style="color:red; background:none">\w</code> и т.д. В обычных строках Python слэш тоже «особенный» (например, <code style="color:red; background:none">\n</code> — перевод строки), поэтому часто используют «сырые» строки с префиксом <code style="color:red; background:none">r</code>.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Запомнить.</b> Для шаблонов регулярных выражений почти всегда пишите их как «сырые строки»: <code style="color:red; background:none">r"\d+"</code>, <code style="color:red; background:none">r"\w+"</code>. Так Python не будет сам интерпретировать слэши, и шаблон останется таким, как вы его задумали. </div>

## Квантификаторы

Управляют количеством повторений элемента:

<code style="color:red; background:none">*</code> — 0 или более раз.

<code style="color:red; background:none">+</code> — 1 или более раз.

<code style="color:red; background:none">?</code> — 0 или 1 раз.

<code style="color:red; background:none">{n}</code> — ровно <code style="color:red; background:none">n</code> повторений.

<code style="color:red; background:none">{n,}</code> — не менее <code style="color:red; background:none">n</code>.

<code style="color:red; background:none">{,m}</code> — не более <code style="color:red; background:none">m</code>.

<code style="color:red; background:none">{n,m}</code> — от <code style="color:red; background:none">n</code> до <code style="color:red; background:none">m</code> включительно.
​

Примеры:

<code style="color:red; background:none">\d+</code> — одно или более чисел (целое число любой длины).

<code style="color:red; background:none">[a-z]{2,4}</code> — от 2 до 4 строчных латинских букв подряд.

<code style="color:red; background:none">\s*</code> — «любое количество» пробелов.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Интуитивно.</b> Читайте <code style="color:red; background:none">+</code> как «один или больше», <code style="color:red; background:none">*</code> как «сколько угодно, даже ноль», а <code style="color:red; background:none">?</code> как «опционально» (может быть, а может и нет). </div>

<hr>

По умолчанию квантификаторы <code style="color:red; background:none">*</code>, <code style="color:red; background:none">+</code>, <code style="color:red; background:none">{n,m}</code> — жадные: они стараются захватить как можно больше символов, при этом всё ещё удовлетворяя шаблону.
​

«Ленивый» вариант добавляет <code style="color:red; background:none">?</code>:

<code style="color:red; background:none">*?</code>

<code style="color:red; background:none">+?</code>

<code style="color:red; background:none">{n,m}?</code>
​

Пример:

<code style="color:red; background:none"><.*></code> — захватит максимально большой фрагмент между первой <code style="color:red; background:none"><</code> и последней <code style="color:red; background:none">></code>.

<code style="color:red; background:none"><.*?></code> — захватит минимальный возможный фрагмент (по одному тегу).
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Практика.</b> Если ваш шаблон «слишком много съедает», попробуйте заменить <code style="color:red; background:none">*</code> или <code style="color:red; background:none">+</code> на <code style="color:red; background:none">*?</code> или <code style="color:red; background:none">+?</code>. </div>

```python
# Неправильная строка, обычная строка и «сырая» строка для одного и того же шаблона
import re
pattern_bad = "\d+"
pattern_normal = "\\d+"
pattern_raw = r"\d+"

print("Неправильная строка:", pattern_bad)
print("Обычная строка:", pattern_normal)
print("Сырая строка:", pattern_raw)

text = "Сегодня 05.03.2026, а вчера было 04.03.2026 2020"

print(re.findall(pattern_bad, text)) # все равно найдет, но будет ругаться на нас
print(re.findall(pattern_normal, text))
print(re.findall(pattern_raw, text))
```

```python
html = "<title>Мой сайт</title><h1>Заголовок</h1>"

greedy = re.findall(r"<.*>", html)
lazy = re.findall(r"<.*?>", html)

print("Жадный вариант:", greedy)
print("Ленивый вариант:", lazy)
```

```python
text = "ID: 123, ID: 456, ID: 789"

print(re.search(r"\d+", text)) # отдаст объект типа Match
print(re.search(r"\d+", text).group()) # чтобы вывести результат, нужно использовать метод group()
```

```python
print("re.findall:")
print(re.findall(r"\d+", text)) # выдаст список
```

```python
text = "ID: 1234, ID: 4564, ID: 7893"

print("re.sub (маскируем ID):")
print(re.sub(r"\d+", "***", text))
```

```python
text = "ID: 123, ID: 456, ID: 789"

print("re.split (разбиваем по запятым и пробелам):")
print(re.split(r"[, ]+", text))
```

```python
text = "Скидки: -5%, -10%, -15%, а иногда -100%!"

print(r"Числа с %, \d+%:") # даже тут мы можем использовать сырые строки, чтобы Python корректно печатал строку
print(re.findall(r"\d+%", text))
```

```python
phones = "Телефоны: +7-999-123-45-67, 8(800)555-35-35, 12345"

print("Последовательности цифр длиной от 2 до 4:")
print(re.findall(r"\d{2,4}", phones))
print(re.findall(r"[0-9]{2,4}", phones))
```

## Якоря: начало и конец строки, границы слов

Якоря задают «позиции» в строке, а не символы:

<code style="color:red; background:none">^</code> — начало строки.

<code style="color:red; background:none">$</code> — конец строки.

<code style="color:red; background:none">\b</code> — граница слова (между «словесным» символом и «не словесным»).
​

<code style="color:red; background:none">\B</code> — «не граница» слова.
​

Примеры:

<code style="color:red; background:none">^Hello</code> — строка начинается с <code style="color:red; background:none">Hello</code>.

<code style="color:red; background:none">world$</code> — строка заканчивается на <code style="color:red; background:none">world</code>.

<code style="color:red; background:none">\bcat\b</code> — слово <code style="color:red; background:none">cat</code> как отдельное слово, но не часть <code style="color:red; background:none">concatenate</code>.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> Выражение <code style="color:red; background:none">^pattern$</code> проверяет, подходит ли <i>вся строка</i> под шаблон. Это удобно для валидации форматов (телефон, email, дата). </div>

```python
text1 = "Hello world"
text2 = "Well, Hello world!"

print("Начинается ли строка с 'Hello'?")
print(bool(re.match(r"^Hello", text1)))
print(bool(re.match(r"^Hello", text2)))
```

```python
text3 = "cat catalog scatter cat."
print("\nСлово 'cat' как отдельное слово:")
print(re.findall(r"\bcat\b", text3))
```

```python
print("Вхождения 'cat' как часть других слов:")
print(re.findall(r"cat", text3))
```

## Группировка

Скобки выполняют две роли:
* Группировка: управляют областью действия квантификаторов и альтернатив.
* Захват: позволяют извлекать части совпадения.

**Примеры группировки:**

<code style="color:red; background:none">(ab)+</code> — подряд одна или более строк <code style="color:red; background:none">ab</code> (например, <code style="color:red; background:none">ababab</code>).

<code style="color:red; background:none">(cat|dog)</code> — либо <code style="color:red; background:none">cat</code>, либо <code style="color:red; background:none">dog</code>.

**Захватывающие группы:**

<code style="color:red; background:none">()</code> создаёт группу;

метод <code style="color:red; background:none">match.group(1)</code> возвращает текст, совпавший с первой группой и т.п.

метод <code style="color:red; background:none">match.group(0)</code> вернет весь найденный шаблон.

метод <code style="color:red; background:none">match.groups()</code> вернет кортеж из найденных групп.
​

**Именованные группы:**

синтаксис: <code style="color:red; background:none">(?P\<name>...)</code>;

доступ: <code style="color:red; background:none">match.group("name")</code>.
​

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Паттерн для даты.</b> Например: <code style="color:red; background:none">r"(?P<day>\d{2})-(?P<month>\d{2})-(?P<year>\d{4})"</code> — даёт три именованные группы, которые легко читать в коде. </div>

```python
date_text = "Дата: 05-03-2026"

pattern = r"(?P<day>\d{2})-(\d{2})-(?P<year>\d{4})"
n = re.search(pattern, date_text)

print(n.group(0))
print(n.group(1))
print(n.group(2))
print(n.group(3))
print(n.group('day'))
```

```python
date_text = "Дата: 05-03-2026"

pattern = r"(?P<day>\d{2})-(?P<month>\d{2})-(?P<year>\d{4})"
m = re.search(pattern, date_text)

print("День:", m.group("day"))
print("Месяц:", m.group("month"))
print("Год:", m.group("year"))
print("Все группы:", m.groups())
```

## Бекреференсы (обратные ссылки)

Бекреференс позволяет повторно сопоставить <b>то же самое содержимое</b>, которое ранее было захвачено группой в круглых скобках.<br>
Сначала вы задаёте группу <code style="color:red; background:none">(...)</code>, а затем ссылаетесь на неё как на уже найденный фрагмент шаблона.

- Захватывающая группа: <code style="color:red; background:none">(...)</code>  
- Ссылка на неё: <code style="color:red; background:none">\1</code>, <code style="color:red; background:none">\2</code>, <code style="color:red; background:none">\3</code> и т.д. — по номеру группы слева направо.

Пример шаблона для повторяющегося слова:<br>
<code style="color:red; background:none">\b(\w+)\b\s+\b\1\b</code><br>
Интуитивно:  
- <code style="color:red; background:none">(\w+)</code> — захватываем слово как группу 1;  
- <code style="color:red; background:none">\1</code> — требуем то же самое слово ещё раз после пробелов.

Такой паттерн найдёт, например, фрагменты <code style="color:red; background:none">is is</code> или <code style="color:red; background:none">test   test</code>.

<p><b>Типичные применения бекреференсов</b></p>

- Повторяющиеся слова в тексте:  
  <code style="color:red; background:none">\b(\w+)\b\s+\b\1\b</code> → находит удвоения вроде <code style="color:red; background:none">this THIS</code> (с флагом <code style="color:red; background:none">re.IGNORECASE</code>).
- Повторяющиеся фрагменты фиксированной формы:  
  <code style="color:red; background:none">([A-Z]\d)\1</code> → совпадения вроде <code style="color:red; background:none">A1A1</code>, но не <code style="color:red; background:none">A1B1</code>.
- Парные кавычки:  
  <code style="color:red; background:none">(['"])(.*?)\1</code> — текст в одинаковых кавычках: <code style="color:red; background:none">'...'</code> или <code style="color:red; background:none">"..."</code>.

<p><b>Важные замечания</b></p>

- Бекреференс сопоставляет <b>буквально тот же текст</b>, который поймала группа, а не «любой текст такой же формы».
- Нумерация относится только к захватывающим группам <code style="color:red; background:none">(...)</code>; незахватывающие группы <code style="color:red; background:none">(?:...)</code> в счёт не идут.
- В Python при сложных шаблонах именованные группы и бекреференсы <code style="color:red; background:none">(?P&lt;name&gt;...)</code> и <code style="color:red; background:none">(?P=name)</code> часто делают выражение гораздо более читаемым.

```python
text = "This is is a test. And this THIS is another test test."

re.findall(r"\b(\w+)\b\s+\b\1\b", text, flags=re.IGNORECASE)
```

## Компиляция шаблонов

Функция <code style="color:red; background:none">re.compile</code> создаёт объект шаблона, который можно использовать много раз.

Преимущества:

* не нужно каждый раз заново создавать строку шаблона;
* код становится более читаемым;
* удобно задавать флаги один раз (например, нечувствительность к регистру).
​

Пример:

<code style="color:red; background:none">pattern = re.compile(r"\d+")</code>

<code style="color:red; background:none">pattern.findall(text)</code>

<div style="background-color: rgba(220,255,5,1.00); padding: 15px; margin: 15px; border-radius: 8px; border-left: 5px solid rgba(220,255,5,1.00);color:black; width:80%;"> <b>Совет.</b> Если вы используете один и тот же шаблон более двух раз, имеет смысл его скомпилировать и хранить в переменной. </div>

```python
logs = [
    "2026-03-02 INFO User logged in, id=123",
    "2026-03-02 ERROR Something went wrong, code=500",
    "2026-03-02 INFO User logged out, id=123",
]

pattern = re.compile(r"\b(INFO|ERROR)\b")

m = pattern.search(logs[0])
print("Уровень:", m.group(), "| строка:", logs[0])

n = pattern.search(logs[1])
print("Уровень:", n.group(), "| строка:", logs[1])

p = pattern.search(logs[2])
print("Уровень:", p.group(), "| строка:", logs[2])
```

## Флаги

Флаги изменяют поведение движка регулярных выражений.

Часто используемые флаги:

<code style="color:red; background:none">re.IGNORECASE</code> или <code style="color:red; background:none">re.I</code> — игнорировать регистр букв.
​

<code style="color:red; background:none">re.MULTILINE</code> или <code style="color:red; background:none">re.M</code> — якоря <code style="color:red; background:none">^</code> и <code style="color:red; background:none">$</code> работают для каждой строки внутри текста.
​

<code style="color:red; background:none">re.DOTALL</code> или <code style="color:red; background:none">re.S</code> — точка <code style="color:red; background:none">.</code> начинает совпадать и с переносами строк.
​

Флаги можно передавать:
* как параметр: <code style="color:red; background:none">re.search(pattern, text, flags=re.I)</code>;
* или при компиляции: <code style="color:red; background:none">re.compile(pattern, re.I | re.M)</code>.
​

```python
text = "Python\nPYTHON\npYtHoN"

print("Игнорируем регистр:")
print(re.findall(r"python", text, flags=re.IGNORECASE))
```

```python
multi_text = "строка1\nстрока2\nстрока3"
print(multi_text, end='\n\n')

print("Поиск строк, начинающихся со 'строка2':")
print(re.findall(r"^строка2", multi_text, flags=re.MULTILINE))
```

```python
dot_text = "Первая строка.\nВторая строка."
print("'.*' без DOTALL:")
print(re.search(r".*", dot_text).group())
```

```python
print("'.*' с DOTALL:")
print(re.search(r".*", dot_text, flags=re.DOTALL).group())
```
